```python
import csv
import shutil
from pathlib import Path
from PIL import Image

# ====== 你需要改的路径 ======
src_dir = Path("/home/erie_lab/Documents/kxz365/ECSE465/25Fall-CSDS-465-Final-Project/leafgan/results/healthy2complex_leafGAN/train_latest/images")
dst_dir = Path("/home/erie_lab/Documents/kxz365/ECSE465/25Fall-CSDS-465-Final-Project/data/generated_complex")
csv_path = Path("/home/erie_lab/Documents/kxz365/ECSE465/25Fall-CSDS-465-Final-Project/data/generated_complex.csv") 
label_name = "complex"                         # 你说所有 label 都是 complex

# 创建输出目录
dst_dir.mkdir(parents=True, exist_ok=True)

rows = [("image", "labels")]  # CSV 表头
counts = 0
for img_path in src_dir.iterdir():
    if not img_path.is_file():
        continue

    if "fake_B" not in img_path.name:
        continue
    
    # 只处理 png / 或其他你想要的格式
    if img_path.suffix.lower() not in [".png", ".jpg", ".jpeg"]:
        continue
    counts += 1
    # 目标 jpg 文件名（去掉 fake_B 也行，这里保留原名只换后缀）
    out_name = img_path.stem + ".jpg"
    out_path = dst_dir / out_name

    # 如果是 png 就转 jpg；如果本来就是 jpg 就直接复制
    if img_path.suffix.lower() == ".png":
        # PNG 可能带透明通道，转 JPG 前先转 RGB
        im = Image.open(img_path).convert("RGB")
        im.save(out_path, quality=95)
    else:
        shutil.copy(img_path, out_path)

    # 记录 CSV 行
    rows.append((out_name, label_name))
    print(f"Processed: {img_path.name} -> {out_name}")

# 写 CSV
with open(csv_path, "a", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

print(f"\nDone! JPGs saved to: {dst_dir}")
print(f"CSV saved to: {csv_path}")
print(f"Total images processed: {counts}")
```
