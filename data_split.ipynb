```python
import os
import shutil
import pandas as pd
from sklearn.model_selection import train_test_split
%load_ext autoreload
%autoreload 2
```

```python
DATA_ROOT = "./data"
TRAIN_IMG_DIR = os.path.join(DATA_ROOT, "train_images")

NEW_TEST_IMG_DIR = os.path.join(DATA_ROOT, "manual_test_images")
os.makedirs(NEW_TEST_IMG_DIR, exist_ok=True)

TRAIN_CSV = os.path.join(DATA_ROOT, "train.csv")

TRAIN_SPLIT_CSV = os.path.join(DATA_ROOT, "train_split.csv")
TEST_SPLIT_CSV  = os.path.join(DATA_ROOT, "test_split.csv")

df = pd.read_csv(TRAIN_CSV)
print(f"Original dataset size: {len(df)}")

train_df, test_df = train_test_split(
    df,
    test_size=0.1,
    random_state=42,
    shuffle=True,
)

print(f"Train: {len(train_df)}, Test: {len(test_df)}")

print("Copying test images...")

for img_name in test_df["image"]:
    src = os.path.join(TRAIN_IMG_DIR, img_name)
    dst = os.path.join(NEW_TEST_IMG_DIR, img_name)
    shutil.copy(src, dst)

print("Image copying done.")


train_df.to_csv(TRAIN_SPLIT_CSV, index=False)
test_df.to_csv(TEST_SPLIT_CSV, index=False)

print(f"train_split.csv saved to: {TRAIN_SPLIT_CSV}")
print(f"test_split.csv saved to:  {TEST_SPLIT_CSV}")
```
